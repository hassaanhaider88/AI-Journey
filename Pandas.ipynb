```python
# Cleaning Data => this is largest job
df.isnull()
df.isnull().sum()
# df2 = df.dropna() # use to drop / delete all null values
# df3 = df.dropna(subset=['city']) # use to drop / delete all null values for a specific columns
# df4 = df.drop(columns=["city"]) # use to drop / delete a complete columns


```

```python
df["age"].fillna(df["age"].mean(),inplace=True)
```

```python
# df["salary"] = df["salary"].str.replace(",","",regex=False)
df["salary"] = pd.to_numeric(df["salary"],errors="coerce")
df['salary'] = df["salary"].fillna(df['salary'].median())
```

```python
df["salary"].isnull().sum()
```

```python
df["department"].value_counts()
```

```python
df['department'] = df["department"].fillna(df['department'].mode()[0])
```

```python

df['age'] = df["age"].fillna(df['age'].median(),inplace=True)
```

```python

df['city'] = df["city"].fillna(df['city'].mode()[0])
```

```python
df
```

```python
df.duplicated().sum()
# df.reset_index()
df.drop_duplicates(subset="employee_id",keep="first")
```

```python
df["annual_salary"] = df["salary"] * 12
```

```python
def categorize(num):
    if num > 5:
        return "Experience Candidate"
    elif num <=5:
        return "Learner"
    else :
        "Invalid"
```

```python
df["experience_status"] = df["experience"].apply(categorize)
df.head()

#with lambda function
df["salary_level"] = df["salary"].apply(lambda salary: "High" if salary > 70000 else "Normal")
df.head()
```

```python
df.sort_values(by="salary")
df.sort_values(by="salary",inplace=True) # this is deep copy (update original data)
df.sort_values(by="salary",ascending=False) # this is shallow copy
```

```python
df["department"] = df["department"].str.strip().str.title()
```

```python
df.sort_values(by = ["department","salary"],ascending=[True,False])
```
