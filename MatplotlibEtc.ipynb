```python
x = np.linspace(0,25,11)
y = x ** 2
```

```python
plt.plot(x,y)
plt.title("Hello Title ")
plt.xlabel("X Axis Yaha pr")
plt.ylabel("Y Axis yaha pr")
```

```python
plt.subplot(2,2,1)
plt.plot(x,x)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.subplot(2,2,2)
```

```python
# distribution plot
import seaborn as sns
import matplotlib.pyplot as plt
```
