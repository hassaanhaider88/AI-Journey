```python
df = df.drop_duplicates()
df.shape
```

```python
df = df[df["Billing Amount"] >= 0]
df
```

```python
df["Name"] = df["Name"].str.title()
df["Name"]
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
df["length to Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
df["length to Stay"]
```

```python
def get_age_group(age):
    if age <= 12:
        return "Childern"
    elif age <= 19:
        return "Teen"
    elif age <=59:
        return "Adult"
    else : 
        return "Senior"
```

```python
df["Age Group"] = df["Age"].apply(get_age_group)
df.head()
```

```python
top10 = df.sort_values(by="Billing Amount",ascending=False).head(10)
top10
```

```python
conditional_summary = df.groupby("Medical Condition").agg(
    average_billing=("Billing Amount", "mean"),
    average_stay=("length to Stay", "mean")
)

conditional_summary
```
